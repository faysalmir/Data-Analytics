## A List is a sequence

```python
# To create a list, enclose elements in square brackets

# list of integers
[10, 20, 30, 40]

# list of strings
['crunchy frog', 'ram bladder', 'sky lark']

# list within a list i.e. a nested list
['spam', 2.0, 5, [10, 20]]
```

```python
# lists can be assigned to variables

cheeses = ['Cheddar', 'Edam', 'Gouda']
numbers = [17, 123]
empty =[]

print(cheeses, numbers, empty)
```

```python
# Lists are mutable

print(cheeses[0])

# printing specific elements of list
numbers = [17, 123]
print(numbers[0])

# Lists are mutable -- replacing sepcific element in the list by an another
numbers[1] =5
print('Lists are mutable- Old list: [17, 123];  New list:', numbers)
```

```python
# The 'in' operator works on lists
'Cheddar' in cheeses

'Brie' in cheeses
```

```python
# Traversing lists
'''
If you want to read the elements of  a list, use the for loop
'''
cheeses = ['Cheddar', 'Edham', 'Gouda']
for cheese in cheeses:
    print(cheese)
```

```python
'''
If you want to perform operations on the members of the list, you need to use indices to make the for-loop run
In this situation, the len() and the range() functions come in handy

''' 
numbers = [17, 123]
for i in range(len(numbers)):        # range(m) creates a list [1,2,3...m]. The len() determines string length 
    numbers[i] = numbers[i] * 2
    print(numbers[i])
    
# Note that the for loop will never execute on an empty list
```

```python
# LIST OPERATIONS
 
a = [1, 2, 3]
b = [4, 5, 6]

# concatenating lists with (+)
c = a + b
print('Concatenated List:' , c)

# repeating lists with (*)

[0] * 4                # This will repeat 0 four times [0, 0, 0, 0], not shown
 

[1,2,3]*3


```

```python
# LIST SLICING

t = ['a', 'b', 'c', 'd', 'e', 'f',]

# here are the slices
t[1:3]
t[:4]
t[3:]
t[:]

# printing output based on above slices
print('t[1:3] is', t[1:3])
print('t[:4]  is', t[:4])
print('t[3:]  is', t[3:])
print('t[:]   is', t[:])


```

```python
# replacing part of a list with another list  
#    slices from a specified element upto but not including the element on the right

t = ['a', 'b', 'c', 'd', 'e', 'f',]
t[1:3] = ['x', 'y']

print('replacing part of a list with another list\n', t)
```

```python
# LIST METHODS
"""
The following methods will be considered:
append()
extend()
sort()
Remember the syntax is object.method(); and lists are objects
"""    
# append() method
t = ['a', 'b', 'c']
t.append('d')
print('Results of append():', t, ' \n')

# extend() takes another list and appends it to the existing one
t1 = ['a', 'b', 'c']
t2 = ['d', 'e']
t1.extend(t2)
print('results of t1.extend(t2):\n', t1, '  <-- note: new elements of t2 are added to t1') 
```

```python
# DELETING ELEMENTS
"""
pop() 
del
remove()
Remember the syntax is: object.method(); and list is an object 
"""

# pop() an item out of the list by index
t = ['a', 'b', 'c']
x= t.pop(1)          # the argument is the index of the element to be removed 
print('list after pop(1): ', t)
print('popped-out item:', x)
```

```python
# Delete() an item from the list by its index
t = ['a', 'b', 'c']
del t[1]
print('list after removing item at index=1:',t, '\n')

# remove() and item from the list (without using index)
t = ['a', 'b', 'c']
t.remove('b')
print("list after removing 'b':", t, '\n')

# Deleting a slice of the list
t= ['a', 'b', 'c', 'e', 'f']
del t[1:5]
print('list after deleting a slice:', t,)
```

```python
# Computing arithmetic mean of user entered numbers
# We will use an infinite while loop, which will stop only when user types 'done' instead of a number
# The while loop will calculate cumulative total and count
# Mean will be determined by dividing cumulative total by final count (when user types 'done')
total = 0
count = 0

while(True):
    inp = input('Enter a number : ')
    if inp == 'done': break
    value = float(inp)              # it is necessary to convert the user input into a floating point number
    total = total + value
    count =  count + 1

average = total / count
print('Average:', average, '\n')
print()
print('sum of squares of deviations from mean, divided by n (length of list)' )
```

```python
# Alternative way of computing mean with list append() function
# We have to make an empty number list first to which we can append numbers

numlist = list()                        # creating an empty list
while (True):
    inp = input('Enter a number: ')
    if inp == 'done': break
    value = float(inp)                  #this line is outside the if-block but inside the while-loop body
    numlist.append(value)

average = sum(numlist)/len(numlist)
print('Average: ', average)
```

# LISTS & STRINGS
## We will learn the following
- Convert a string into a list with list()
- Split a string into a list of words using space as delimiter using split()
- Split a string into a list using a given delimiter using split(delimiter)
- Reverse the process -- i.e. convert a list back into a string using join()

```python
# converting a string into a list
s = 'spam'
t = list(s)
print('Converting string to list: ', t)

# converting a string into a list of words (space as delimiter)
s = 'Jashn e Azadi mubarik'
t = s.split()
print('print slogan as list: ', t)
print('Second element of list i.e. t[2]: ' ,t[2])
print()

# using delimiter (-) to split the string into a list
s = 'spam-spam-spam'
t = s.split('-')
print("Splitting list using delimiter '-': ", t)
```

```python
# join() method concatenates a list of strings
#    join() is a list method, so you have to invoke it on a delimiter and pass the list as a parameter

t = ['Jashn', 'e', 'Azadi', 'mubarik']
delimiter = ' '                                # declaring space as a delimiter
s = delimiter.join(t)
print('Joined with space as delimiter:', s)

# Try using dash ('-') as a delimiter
```

# Parsing lines

```python
"""
- Sometimes we want to find a particular line in text and then find a word in it
- We will use startswith() to indentify the line
- Then list method split() will be used to create a list of words
- Finally, the required word will be printed using index of the list
"""

fhand = open('mbox-short.txt')
for line in fhand:
    line = line.rstrip()              # strips all spaces on the right
    if not line.startswith('From '): 
        continue                      # remaining lines processed only for the 'From'-line
    words = line.split()              # creating a list using space as delimiter
    print(words[2])                   # printing day-of-week using list index      
    
```

# List arguments

```python
'''
Functions can have lists as arguments. 
The function delete_head() below deletes the first element of the list 
'''

# defining the function
def delete_head(t):          # this function works for an arbitray t where t is a list
    del t[0]
    
# calling the function
letters = ['a', 'b', 'c']
delete_head(letters)         # when we make the function call we use list name as argument 
print(letters)
    
```

```python
# We must distinguish between operations that modify lsits and those that create new lists
# append() modifies a list. Concatenate (with '+' sign) creates a new list

t1 = [1, 2]
t2 = t1.append(3)
print('Printing t1:', t1)
print('Printing t2:', t2)   

# Concatenating lists creates new list
t3 = t1 + [3]
print('Printing t3:', t3)
print()
print('Test t2 is t3:\n') 
t2 is t3
```
