# Tuples are immutable

```python
"""
Definition:
A tuple is a sequence of values -- like a list -- but unlike a list
the values are *immutable*. Tuples are comparable and hashable. So 
we can use sort lists of tuplesand use tuples as keys in dictionaries
"""
# A tuple is a comma separated list 
t = 'a', 'b', 'c', 'd', 'e'

# to help identify tuples quickly, use parentheses
t = ('a', 'b', 'c', 'd', 'e')

#  A single element tuple must contain a final comma
t1 =('a',)
print('What is t1?', type(t1))

#  Without the final coma, Python considers it a string
t2 = ('a')
print('What is t2?', type(t2))

# Built-in function tuple() can also be used
t = tuple()
print('Empty tuple using built-in function:', t)
```

```python
"""
If the argument is a sequence (string, list or tuple) then
the function tuple() returns a tuple with elements of 
the sequence
""" 
t = tuple('lupins')
print(t)
```

```python
# Most list operators work on tuples
t = ('a', 'b', 'c', 'd', 'e')
t[0]
```

```python
t[1:3]
```

```python
"""
However, you cannot replace an element because 
tuples are immutable. Trying to replace
the first element - a - with 'A' does not work
"""
t[0] = 'A'
```

```python
"""
Tuples can be concatenated, which provides an indirect way 
to make element replacements in tuples, as shown below:
""" 
t = ('A', ) + t[1:]
print(t)


```

## Comparing tuples

The comparison is done element-wise.<br>
> Starting with the first elements of both tuples, a comparison is made. <br> If they are the equal, the second  elements are compared, <br> until elements that differ are found.<br>
<b> <font color=purple> The subsequent elements are NOT considered </font>




```python
(0, 1, 2) < (1, 3, 4)
```

```python
"""
The following comparison is True. 
This is because its the 2nd element which differs
So the 3rd is not considered 
"""
(0, 1, 200000) < (0, 3, 4)
```

## Sorting tuples
Sort function works on tuples.<br>
It sorts by first element<br>
In case of a tie it sorts by second element, and so on<be>

### __Decorate-Sort-Undecorate__
__Decorate__ a sequence of tuples alogwith sort keys preceeding sequence elements<br>
__Sort__ the sequence of tuples (_by first elements i.e. sort keys_)<br>
__Undecorate__ by getting rid of sort keys adn extracting the sequence<br>
    
__EXAMPLE:__
Suppose you want to sort of list of words ___by length.___
We do it in two steps below:
1. The first loop below builds a list t of tuples where each tuple is ___(l, word)___   i.e word preceded by word-length<br>
    - Sort compares tuples -- first, the lengths (then the words if there is a tie).<br>
    - Sorting is in reverse order i.e longer words first

2. The second loop traverses the list of tuples and build a list (nammed res) of words.<br>
They are in reverse order of length 



```python
txt = 'but soft what light in yonder window breaks'

#Splitting text into words (using space as delimiter)
words = txt.split()

# creating a list of tuples consiting of lenghts, words
t = list()
for word in words:
    t.append((len(word), word))

# Sorting tuples (in reverse order)
#    Recall tuples sorted by first elements unless there is a tie
t.sort(reverse=True)
print('List t  :', t)

#Putting the tuples back into a list of words
res = list()
for length, word in t:     #iteration over length AND word (i.e. over by-groups)
    res.append(word)

print('List res:', res)

```

# Dictionaries and tuples
Dictionaries have a method called <font comic color = blue> __items__ </font> that returns tuples (from the dictionary)
where each tuple is a ___key-value pair___

___A dictionary can be thought of as a collection of tuples of size 2___
    

```python
d ={'a':10, 'b':1, 'c':22}
t=list(d.items())
print(t)               #dictionary may not be in any particular order
                       # in which case, sort the list created above

# you may want to sort this list
t.sort()
print('Sorted list:', t)
```

# Multiple assignments with dictionaries

```python
"""
The items method can be combined with for-loop to traverse 
the key-value pairs in a dictionary
"""

for key, value in list(d.items()):
    print(value, key)               # first item in the pair is deliberately
                                    # a value, not a key
print('The above pairs are sorted by keys: a, b, c; not by values: 10,1,22')
```

```python
# Sorting a dictionary by values instead of keys
"""
The above reversal of key-value order is useful for sorting dictionary
by values rather than by keys. The trick (as shown above) is to create 
tuples where the value is the first item in the tuple.
"""
l=list()
for key, value in list(d.items()):
    l.append((value, key))          # tuples with value as 1st element
print('Unsorted list:', l)    

l.sort(reverse=True)
print('Sorted list  :', l)    
```
