# Updating Variables

```python
##    Updating variables
'''Sometimes we want to increment the value of a variable (say, by 1) and save the result in the same variable.
    Doing x = x + 1 will not work because Python evaluates RHS first where it would not find a value for x.
    The solution is to INITIALIZE x first. This can be done by etting x equal to some value. eg. x = 0.
'''

x = 0
x = x + 1
print(x)

## The resulting value of x (i.e. 1) is printed
```

# The while statement



```python
"""
    The while statement is a form of iteration
    The following  program that starts a countdown from 5 and prints "Blastoff!" when it reaches zero.
    The iteration variable n needs to be initialized at 5 to begin the countdown

    Note that the while-statement below is the start of the while-loop.
    The indented statements are the body of the loop. This is where the iteration variable is decremented by 1,
    and its value is printed.
    
    When the while-condition (n > 0) fails, we jump out of the loop, and the subsequent statement(s) is printed
    which in this case is: print('Blastoff!')

"""

n = 5
while n > 0:
    print(n)
    n = n - 1
print('Blastoff!')


```

# Infinite loops

##  Directions on shampoo: Lather, rise, repeat
## That's actually an infinite loop, no kidding!



# Here's an example of infinite loop
# Please don't make the mistake of actually running it

'''
In the loop below, the logical expression on the while-statement is simply the logical constant True
So we never jump out of the loop because this expression is never false
'''
n=10
while True:
    print(n, end = '  ')
    n = n - 1
print('Done!')

# Even though the iteration variable is initialized at 10, and is decremented by 1. The while-statement is never false
# So we never exit the loop

### If you do not want this program to continue chasing its tail .....  use break

```python
'''
The above loop is dysfunctional but we can write infinite loops that work well.
We use (potentially) infinite loops when we do not know in adance how many times
the loop needs to iterate. So we write a loop that could iterate as many times
as needed but would end if certain condition (eg. a certain special input from the user) is met.
'''

while True:
    line = input('>  ')
    if line == 'done':
        break
    print(line)
print('Done!')
```

# Finishing Iteration with continue

```python
'''
Sometimes you want to not jump to the next iteration before finishing the current one, skipping the
remaining statements in the body of the loop. In this case you can use continue.
In the following example, the user is entering a line, which is to be printed. However, if he wants to write a 
comment starting with a # sign then we want to skip the print statement and go to next iteration.
Alternatively, if the user provdes input 'done' we want to exit the loop. To do this, we use a break.
'''

while True:
    line = input ('>  ')
    if line[0] == '#':
        continue
    if line == 'done':
        break
    
    print(line)
print('Done!')
    
    
```

# Definite loops using for

### also known as "for loops"

```python
'''
When you want the loop to loop through a known set of things, such as a list, use the for loop.
One example of such as list or set is the set of days of a week.
'''
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
for day in days:
    print('Today is ', day)
print('These are the days of our lives!!')

# Question: what will happen if you use days instead of day in the print command?
 
```

# Counting and summing loop

```python
# Counting 

count = 0
for itervar in [3, 41, 12, 9, 74, 15]:
    count = count + 1
print('Count:  ', count)



# Summing

total = 0
for itervar in [3, 41, 12, 9, 74, 15]:
    total = total + itervar
print('Total:   ', total)

# Note that only the last total is printed and not the running total.
# How should you change the above loop to get the running total?
# (Hint: make print() execute inside body of the loop)

# Running sum

total = 0
for itervar in [3, 41, 12, 9, 74, 15]:
    total = total + itervar
    print(total)

# In practice, you can use the len() and sum() function in Python to count and find total, respectively.
```

# Maximum and minimum loops

```python
# Maximum
'''
To find maximum of a list, we initialize a variable largest at None, i.e. empty.
We iterate over the list an compare each number in it with the value of largest.
We replace the largest with the number from list if it is greater than largest 
'''

largest = None
for itervar in [3, 41, 12, 9, 74, 15]:
    if largest is None or itervar > largest:   #Note the syntax --- if largest is None.  largest = None will give error
        largest = itervar
        print ('Loop: ', itervar, largest)
print('Largest: ' , largest)

```
