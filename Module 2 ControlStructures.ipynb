```python
# Module 2: Python Control Structures

# Conditional Execution

'''
The general form of the conditional statement is as follows.

if condition:
    executable statement
    executable statement

The 'condition' after the 'if' is evaluated as True or False, and the
indented statements are executed if the condition is true, otherwise skipped

Note: The if statements have the same structure as function definitions or for loops. 
'''
## Selection

var = -1
if var < 0:
    print(var)
    print ("the value of var is negative")
    
    
```

```python
'''
Sometimes you want to a place holder for executable statements (code that you haven't written yet)
In that case you can use  a pass statement
''' 
x = -1
if x < 0:
    pass
```

# Alternative execution: if-then-else logic

```python
# Sometimes there are two possibilities and the condition determines, which one gets executed
# The alternatives are called branches

x=7
if x%2 == 0 :                          #if the remainder after division by 2 is 0
    print('x is even')
else :
    print('x is odd')
```

# Chained conditionals


## Sometimes there are more than two possibilities and we need more than two branches


```python
x = 1 
y = 1

if x < y:
    print('x is less than y')
elif x > y:                                      #elif stands for 'else if'
    print('x is greater than y')
else:
    print('x is equal to y')
    
print('Now try different values of x and y and see what happens')
```

```python
## There are several variations on teh theme.
## For example, the 'else' statement must come at the end but there does not have to be an else statement

choice = 'b'

if choice == 'a':
    print('Bad guess')
elif choice == 'b':
    print('no prizes for guessing')
elif choice == 'c':
    print('close but not correct')
```

# Nested conditionals

```python
## Sometimes we can nest a condition within another condition.

x = 1; y = 0

if x == y:
    print('x and y are equal')
else :
    if x < y:
        print('x is less than y')
    else:
        print('x is greater than y')

print()      
print('Now try different pairs of x and y values and see what happens')
        
```

```python
# Nested conditionals can become dufficult to read pretty quickly
# An alternative is to use and operator, and avoid the nested conditional

x=2

if 0 < x:
    if x < 10:
        print('x is a positive single-digit number!')
        
```

```python
# The same condition can be written using an 'and' operator

if 0 < x and x < 10:
    print('x is a positive AND single-digit number')
```

```python
# If we give an invalid input, we run into problems
# for example when prompted for input if instead of entering a valid number, we type 'fred', we get an error

inp = input('Enter Farenheit Temperature: ')
fahr = float(inp)
cel = (fahr -32.0) * 5.0/9.0
print(cel)
```

```python
# A way around this difficulty is 'try' and 'except' features in Python

inp = input('Enter Farenheit Temperature: ')
try:                                               # The try block executes if input is valid (i.e. no exception occurs)
    fahr = float(inp)
    cel = (fahr -32.0) * 5.0/9.0
    print(cel)
except:                                            # The except block executes if an exception occurs (due to invalid input)
    print()                                        # either way, the program does not crash.
    print('Read PYTHON FOR DUMMIES!!!')
```
