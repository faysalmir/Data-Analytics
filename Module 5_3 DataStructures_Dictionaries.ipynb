# Module5_3_DataStructures_Dictoinaries

```python
'''
A dictionary is like a list. But the indexdoes not have to be an integer; 
it can be of another type.
Dictionary is a mapping between a set of indices(called keys) and a set of values.
An association of key-value pair is called an item 
'''

# Here's and examples of an English to Urdu dictionary 
#    Items are enclosed in curly brackets; keys and values
#        are separated by colon(:)
eng2urdu ={'one':'aik', 'two':'do', 'three':'teen', 'four':'char', 'five':'paanch'}
print('Printing whole dictionary: \n', eng2urdu)

print()

# You can print values for specific keys
print("Printing item for key='two' \n", eng2urdu['two'])

# length of dictionary
print('length of dictionary: ')
len(eng2urdu)
```

```python
"""
To ckeck if a given key or a value is in the dictionary, you can
use the 'in' operator
"""

print("Is the key 'one' in dictionary")
'one' in eng2urdu

```

```python
# the 'in' option does not work on values
print("Is the value 'aik' in dictionary?")
'aik' in eng2urdu
      
# For checking specific values you have to first list the values 
# using values() function and then check by using use 'in'
vals = list(eng2urdu.values())
'aik' in vals


```

# Dictionary as a set of counters

```python
# Suppose you wat to cont how many times each letter in  a string appears.
# You can create a dictionary that calculates the frequencies of words in the string
# here's how to create such a dictionary

word = 'Nanga Parbat'
d = dict()            # create empty dictionary
for c in word:
    if c not in d:
        d[c] = 1
    else:
        d[c] = d[c] + 1
print(d)

# How would you modify the program to accept word supplied by the user?
```

```python
inp = input('Type word here: ')
word = inp
d = dict()            # create empty dictionary
for c in word:
    if c not in d:
        d[c] = 1
    else:
        d[c] = d[c] + 1
print(d)
```

```python
# Dictionary method get() takes a key and a default value.
# If the key is in the dictionary, get() returns the corresponding value
# else itreturns the default value
# This is used to write better loop for generating freq dist of letters in a word

word = 'Nanga Parbat'
d = dict()
for c in word:
    d[c] = d.get(c, 0) + 1
print(d)
```

# Dictionaries and files

```python
fname = input('Enter the file name: ')
try:
    fhand = open(fname)
except:
    print('File cannot be opened: ', fname)
    exit()

counts = dict()
for line in fhand:
    words = line.split()   #splitting words using space as delimiter
    for word in words:
        if word not in counts:
            counts[word] = 1
        else:
            counts[words] += 1
print(counts)
```

# Looping and dictionaries

```python
# You can use the keys of a dictionary to iterate on
# Below a dictionary named counts is created.
# Its keys are used as iteration variable

counts = {'chuck': 1, 'annie':42, 'jan': 100}   #creating dictionary

# using dictionary keys as iteration variable
for key in counts:                             
    print(key, counts[key])
```

```python
# Find all entities in a dictionary with values above 10

counts = {'chuck': 1, 'annie': 42, 'jan': 100}
for key in counts:
          if counts[key] > 10 :
            print(key, counts[key])
```

```python
# Sort the dictionary keys in alphabetical order
#    >Use keys() method in dictionary object to list the dictionary keys
#    >Sort the keys list (IMPORTANT: we are not sorting the dictionary!)
#    >Print sorted keys and corresponding values 

# creating a list of keys using keys() method
counts = {'chuck': 1, 'annie': 42, 'jan': 100}
lst = list(counts.keys())                     
print('Unsorted list of keys:', lst)
print()

#sorting keys list 
lst.sort()                        

## call the sorted keys via a for-loop, print key-value pairs
## for loop on sorted list of keys 

for key in lst:
    print(key, counts[key])
 
```
